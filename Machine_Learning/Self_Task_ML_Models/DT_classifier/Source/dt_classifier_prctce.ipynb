```python
df["Education"] = df["Education"].fillna(df["Education"].mode()[0])
df["Education"]
```

```python
df["Income"] = df["Income"].fillna(df["Income"].mean())
df["Income"] = df["Income"].round(2)
```

```python
df["Married"] = df["Married"].fillna(df["Married"].mode()[0])
df["Married"]
```

```python
print(df)
```

```python
df.isnull().sum()
```

##### Feature and target

```python
# Feature selection
X = df[[
    "Age",
    "Income",
    "Education",
    "Married"
]]

# Target 
y = df["Default"]
```

```python
pd.concat([X,y],axis=1)
```
