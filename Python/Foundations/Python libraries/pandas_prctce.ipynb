```python
# It filters the DataFrame and returns students from either CSE or IT.
result=df[
    (df["Department"]=="CSE") |
    (df["Department"]=="IT")
    ]
print(result)
```

```python
# 1. find students with attendance >= 75
# 2. find students whose attendance is atleast 75 and python mark is atleast 80
# 3. calculate the average of python, sql and excel

# 1. attendance >= 75
cond=df[df["Attendance"]>=75].columns
cond
```

```python
# 1. find Students with attendance >=75
df[(df['Attendance'] >= 75)]
```

```python
#2. find students whose attendance is at least 75 AND Python mark is at least 80
df[(df['Attendance'] > 75) & (df['Python'] >= 80)][['Attendance','Python']]
```

```python
# 3. Calculate the avg
(df["Python"]+df["SQL"]+df["Excel"])/3
(round(df["Python"]+df["SQL"]+df["Excel"])/3)
```

```python
df["Average"]=((df["Python"]+df["SQL"]+df["Excel"])//3)
df
```

```python
# to store values in new column results 
# Comprehension - means write a single line code 
# so put the "pass"  in first and then condition


# df["Results"] → Creates a new column named "Results".
# Why? → To store the Pass/Fail result.

# = → Assigns the result to the "Results" column.
# Why? → To save the calculated values.

# df["Average"] → Selects the "Average" column.
# Why? → We need the average marks to check Pass/Fail.

# .apply() → Applies a function to each value in the column.
# Why? → To check each student's average individually.

# lambda avg: → Creates a small function.
# Why? → To perform the Pass/Fail check without defining a separate function.

# avg → Represents one student's average value at a time.

# "Pass" if avg >= 50 → Returns "Pass" when average is 50 or more.
# Why? → 50 is considered the passing mark.

# else "Fail" → Returns "Fail" when the average is below 50.
# Why? → The student has not reached the passing mark.


df["Results"]=df["Average"].apply( lambda avg:
   "Pass" if avg>=50 else "Faill"
)
df
```

### Sorting

```python
# Highest -  performing students (based on avg)
# here default  is ascending

df.sort_values(by="Average")
```

```python
# for descending (based on attendance)
df.sort_values(by="Attendance",ascending=False)
```

# GROUP BY

```python
# Number of students per dept
df.groupby("Department")["Student_ID"].count()
```
