```python
# Read image in BGR (default OpenCV format)
img_bgr = cv2.imread("Puppy.jpg")

# Convert to RGB for Matplotlib
img_rgb = cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGB)

# Show side by side
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_bgr)
plt.title("BGR (OpenCV default)")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img_rgb)
plt.title("RGB (Converted)")
plt.axis("off")

plt.show()
```

```python
import cv2

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    'haarcascade_frontalface_default.xml'
)

img = cv2.imread("groupimg.jpg")

gray = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2GRAY
)

faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.2,
    minNeighbors=6

)

for (x,y,w,h) in faces:

    cv2.rectangle(
         img,
         (x,y),
         (x+w,y+h),
         (0,255,0),
         3
     )

    cv2_imshow(img)

    print("Faces detected:", len(faces))
```
